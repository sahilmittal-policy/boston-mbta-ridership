```python
# Load the raw survey files
import pandas as pd
from pathlib import Path

raw_file1 = "/Users/eshagupta/Downloads/CMU/Fall'26/Python/Final Project/MBTA-Passenger_Survey-Line_Values.csv"
raw_file2 = "/Users/eshagupta/Downloads/CMU/Fall'26/Python/Final Project/MBTA-Passenger_Survey-Line_Pct.csv"

ps17_raw = pd.read_csv(raw_file1)
ps17_pct = pd.read_csv(raw_file2)


# Standardize column names
ps17_raw.columns = (
    ps17_raw.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

ps17_pct.columns = (
    ps17_pct.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)


# Inspect the files before filtering
print("Line values:", ps17_raw.shape)
print("Line percentages:", ps17_pct.shape)

print("Modes in values file:")
print(ps17_raw["mode"].value_counts(dropna=False))

print("Lines in values file:")
print(ps17_raw["line"].value_counts(dropna=False).head(40))

print("Modes in percentages file:")
print(ps17_pct["mode"].value_counts(dropna=False))

print("Lines in percentages file:")
print(ps17_pct["line"].value_counts(dropna=False).head(40))


# Keep bus records
ps17_raw["mode"] = ps17_raw["mode"].astype("string").str.strip()
ps17_pct["mode"] = ps17_pct["mode"].astype("string").str.strip()

ps17_raw = ps17_raw.loc[
    ps17_raw["mode"].str.casefold().eq("bus").fillna(False)
].copy()

ps17_pct = ps17_pct.loc[
    ps17_pct["mode"].str.casefold().eq("bus").fillna(False)
].copy()


# Drop English-comprehension fields if they are not needed
ps17_raw = ps17_raw.drop(columns=["ability_to_understand_english_often", 
"ability_to_understand_english_sometimes", 
"ability_to_understand_english_always", 
"ability_to_understand_english_never", 
"ability_to_understand_english_prefer_not_to_say", 
"count_ability_to_understand_english"])
ps17_pct = ps17_pct.drop(columns=["ability_to_understand_english_often", 
"ability_to_understand_english_sometimes", 
"ability_to_understand_english_always", 
"ability_to_understand_english_never", 
"ability_to_understand_english_prefer_not_to_say", 
"count_ability_to_understand_english"])

# Check the cleaned bus tables
print("Bus line values:", ps17_raw.shape)
print("Bus line percentages:", ps17_pct.shape)

display(ps17_raw.head())
display(ps17_pct.head())


# Save the cleaned tables separately
project_root = Path.cwd()

while (
    project_root != project_root.parent
    and not (project_root / "data" / "processed").is_dir()
):
    project_root = project_root.parent

processed_dir = project_root / "data" / "processed"

if not processed_dir.is_dir():
    raise FileNotFoundError(
        "Could not find the project's data/processed folder. "
        "Open the project root folder in VS Code, then rerun this cell."
    )

ps17_raw.to_csv(
    processed_dir / "passenger_survey_bus_line_values_clean.csv",
    index=False,
)

ps17_pct.to_csv(
    processed_dir / "passenger_survey_bus_line_percentages_clean.csv",
    index=False,
)
```
